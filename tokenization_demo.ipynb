```python
!pip install -q nltk
import nltk, re
for pkg in ['stopwords', 'wordnet', 'omw-1.4']:
    nltk.download(pkg, quiet=True)
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
```

## 1. Raw review

```python
review = "It's good to have maxi in downtown but this store is really small in comparison to other outlets."
print(review)
```

## 2. Lowercase (so that words with different capitalization are treated consistently during subsequent processing)

Prompt: "can you please lowercase the review"

```python
lower = review.lower()
print(lower)
```

## 3. Tokenize

Prompt: "Can you tokenize the review now"

```python
tokens = re.findall(r"[a-z]+(?:'[a-z]+)?", lower)
print(tokens)
print(len(tokens), 'tokens')
```

## 4. Remove stopwords

```python
stops = set(stopwords.words('english'))
content = [t for t in tokens if t not in stops]
print(content)
```

## 5. Stemming vs lemmatization

Prompt: "Finally can you show, what stemming and lemmatization do"

```python
stemmer, lemm = PorterStemmer(), WordNetLemmatizer()
stems = [stemmer.stem(t) for t in content]
lemmas = [lemm.lemmatize(t) for t in content]
print('Stemmed:   ', stems)
print('Lemmatized:', lemmas)
```
